```python
from pathlib import Path
import pandas as pd

run_dir = Path("../outputs/runs/2026-02-20_1530")   # <- adjust if needed
signals_path = run_dir / "signals.csv"

df = pd.read_csv(signals_path)
print("signals rows:", len(df))
print("\ncolumns:", df.columns.tolist())
print("\nnull rates (top):")
print(df.isna().mean().sort_values(ascending=False).head(10))
print("\nsample rows:")
print(df.head(5).to_string())
```

```python
import pandas as pd

df = pd.read_csv("../outputs/runs/2026-02-20_1530/signals.csv")

# What sources/metrics are missing values?
print("null value rows:", df["value"].isna().sum(), "of", len(df))

# If you have a 'source' or 'source_table' column, use it:
src_col = "source" if "source" in df.columns else ("source_table" if "source_table" in df.columns else None)
if src_col:
    print("\nNull value rate by source:")
    print(df.groupby(src_col)["value"].apply(lambda s: s.isna().mean()).sort_values(ascending=False).head(20))

print("\nTop metrics with null value counts:")
print(df[df["value"].isna()].groupby("metric").size().sort_values(ascending=False).head(30))
```
